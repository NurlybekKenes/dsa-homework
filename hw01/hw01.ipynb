````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
cities_clean = [i.strip().title() for i in cities] # можно также использовать i.strip().capitalize()
amounts_int = [int(i) for i in amounts]
orders_count = len(amounts_int)
orders_total = sum(amounts_int)
orders_avg = orders_total // orders_count
print(f"Общее число заказов: {orders_count}, Общая сумма: {orders_total}, Средний чек: {orders_avg}")
orders_not_paid = list(filter(lambda x: x[1]==False, zip(amounts_int, paid)))
not_paid_sum = sum(list(map(lambda x: x[0], orders_not_paid)))
print(f"Заказов не оплачено: {len(orders_not_paid)}, Сумма: {not_paid_sum}")
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {is_paid}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
almaty_revenue = 0
astana_revenue = 0
for c, a, p in zip(cities_clean, amounts_int, paid):
    if p and c == "Almaty":
        almaty_revenue += a
    if p and c == "Astana":
        astana_revenue += a
print(f"Выручка по Алматы: {almaty_revenue} ₸, по Астане: {astana_revenue} ₸")

paid_amounts = [a if p else -1 for a, p in zip(amounts_int, paid)]
max_val = max(paid_amounts)
best_index = paid_amounts.index(max_val)
best_city = cities_clean[best_index]
print(f"Номер крупного оплаченного заказа: {best_index}, Город: {best_city}")

weekend_percent = (sum(1 for d in days if d in ["сб", "вс"]) / len(days)) * 100
print(f"Доля заказов выходного дня(сб, вс): {weekend_percent:.2f}%")

# Копилка
target = 400000
deposit = 15000
add_deposit = 1000
saved = 0
saved_with_add = 0
months = 0
months_with_add = 0

while saved < target:
    saved += 15000
    months += 1
    if saved_with_add < target:
        saved_with_add += deposit
        deposit += add_deposit
        months_with_add += 1

print("Задача «копилка»")
print(f"Через сколько месяцев студент накопит на курс: {months} месяцев")
print(f"Если каждый месяц откладывать на 1 000 ₸ больше: {months_with_add} месяцев")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
unique_cities = [] 
revenue_by_city = []
            
for c, a, p in zip(cities_clean, amounts_int, paid):
    revenue = a if p else 0
    
    if c in unique_cities:
        revenue_by_city[unique_cities.index(c)] += revenue
    else:
        unique_cities.append(c)
        revenue_by_city.append(revenue)

sorted_cities_with_revenue = sorted(zip(unique_cities, revenue_by_city), key=lambda x: x[1], reverse=True)

unique_cities = [city for city, _ in sorted_cities_with_revenue]
revenue_by_city = [revenue for _, revenue in sorted_cities_with_revenue]

print("unique_cities: ", unique_cities)
print("revenue_by_city: ", revenue_by_city)

sorted_amounts = sorted(amounts_int)
mid = len(sorted_amounts) // 2

median = (sorted_amounts[mid - 1] + sorted_amounts[mid]) / 2

print("Медиана суммы заказа: ", median)
````
